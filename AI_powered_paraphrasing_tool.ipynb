# 2. Model Selection and Loading

In this project, the **FLAN-T5 small** transformer model was selected for the paraphrasing task.

### Why FLAN-T5 Small?

FLAN-T5 is an instruction-tuned version of the T5 (Text-to-Text Transfer Transformer) model. It is designed to understand and perform tasks based on natural language instructions such as "Paraphrase:" or "Rewrite this sentence".

The following reasons justify the selection of this model:

- **Instruction-based Learning:**  
  FLAN-T5 performs well on tasks like paraphrasing when given clear prompts, making it suitable for this application.

- **Lightweight and Efficient:**  
  The "small" version requires less computational power and memory, making it ideal for execution in environments such as Jupyter Notebook and Google Colab.

- **Ease of Integration:**  
  It can be easily implemented using the Hugging Face Transformers library without requiring complex fine-tuning.

### Why Not Other Models?

Other transformer models were considered but not used for the following reasons:

- **Large Models (e.g., larger T5 variants):**  
  Require high computational resources and may cause memory issues or slow performance.

- **API-based Models (e.g., GPT via external APIs):**  
  Require API keys and internet dependency, which may not align with the requirement of a self-contained console-based application.

- **Traditional Models (e.g., basic NLP techniques):**  
  Do not provide the same level of contextual understanding and paraphrasing capability as transformer-based models.

### Model Loading

The model is loaded using the Hugging Face Transformers pipeline for text-to-text generation, enabling direct input-output transformation for paraphrasing tasks.

```python

# !pip install transformers==4.41.2 sentencepiece language-tool-python -q

from transformers import pipeline
import language_tool_python

# Load model
paraphraser = pipeline("text2text-generation", model="google/flan-t5-small")

# Grammar tool
tool = language_tool_python.LanguageTool('en-US')


# -------------------------------
# FINAL FUNCTION
# -------------------------------
def generate_paraphrase(text):
    result = paraphraser(
        "Paraphrase: " + text,
        max_length=120,
        do_sample=False   # stable + clean
    )

    return result[0]['generated_text']


# -------------------------------
# LONG SENTENCE TEST
# -------------------------------
text = """Artificial Intelligence is rapidly transforming multiple industries across the globe by enabling machines
to analyze large volumes of data, identify patterns, and make intelligent decisions with minimal human intervention."""

print("----- ORIGINAL TEXT -----")
print(text)

paraphrased = generate_paraphrase(text)

print("\n----- PARAPHRASED TEXT -----")
print(paraphrased)


# -------------------------------
# GRAMMAR CHECK
# -------------------------------
matches = tool.check(paraphrased)

print("\n----- GRAMMAR CHECK -----")

if len(matches) == 0:
    print("No issues found")
else:
    for m in matches:
        print(m)

corrected = language_tool_python.utils.correct(paraphrased, matches)

print("\n----- CORRECTED TEXT -----")
print(corrected)
```

#4. Grammar and Spelling Error Detection

To ensure the quality and correctness of the generated paraphrased text, a grammar and spelling checking mechanism is integrated into the system using the LanguageTool library.

### How Errors are Detected

The system uses a rule-based approach provided by the LanguageTool engine:

- The paraphrased text is passed to the grammar checking tool.
- The tool analyzes the sentence based on predefined linguistic rules, including grammar, punctuation, and spelling.
- It identifies errors such as:
  - Incorrect verb forms (e.g., "He go" instead of "He goes")
  - Subject-verb agreement issues
  - Incorrect tense usage
  - Missing punctuation (e.g., commas)
  - Spelling mistakes

### Error Identification Process

- The tool returns a list of detected issues called "matches".
- Each match contains:
  - The type of error
  - The position of the error in the sentence
  - A description of the issue
  - Suggested corrections

### Error Correction

- The detected issues are processed using a correction function.
- Suggested corrections are automatically applied to the paraphrased text.
- The final corrected output is displayed to ensure grammatical accuracy and fluency.

- **Correction Result:**

This approach ensures that the final output is not only semantically meaningful but also grammatically correct and fluent, enhancing the overall quality of the paraphrasing system.

#6. Evaluation Metrics

To evaluate the performance of the paraphrasing system, multiple metrics were used.

### BLEU and ROUGE Scores
The BLEU and ROUGE scores were observed to be very low. This is because the model significantly modifies or simplifies the original sentence, resulting in minimal word overlap.

###CASE 1. Semantic Similarity
To better evaluate meaning preservation, semantic similarity was calculated using TF-IDF vectorization and cosine similarity.

- **Original:**  
She go to school everyday and don't likes homework.

- **Paraphrased:**  
She doesn't like homework.

- **Semantic Similarity Score:** 0.2061  

The score indicates that the paraphrased sentence retains only part of the original meaning. While the main idea (disliking homework) is preserved, additional context (attending school daily) is lost.

### Overall Analysis
The evaluation shows that the model tends to simplify sentences rather than generate detailed paraphrases. This results in low lexical similarity but partial semantic preservation.

This highlights a key limitation of lightweight transformer models, where there is a trade-off between simplicity and completeness of meaning.

```python
# Install lightweight library
!pip install scikit-learn -q

# -------------------------------
# Imports
# -------------------------------
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# -------------------------------
# Example (use ONE case)
# -------------------------------
original = "She go to school everyday and don't likes homework."
paraphrased_text = "She doesn't like homework."

print("----- SEMANTIC SIMILARITY -----\n")

print("Original:", original)
print("Paraphrased:", paraphrased_text)

# -------------------------------
# TF-IDF Vectorization
# -------------------------------
vectorizer = TfidfVectorizer()

vectors = vectorizer.fit_transform([original, paraphrased_text])

# -------------------------------
# Cosine Similarity
# -------------------------------
similarity = cosine_similarity(vectors[0], vectors[1])

print(f"\nSemantic Similarity Score: {similarity[0][0]:.4f}")
```

# 7. Challenges Faced

During the development of the paraphrasing system, several practical and technical challenges were encountered.

### Initial Model Selection Issues

The project initially started with the T5-small model for paraphrasing. However, the model often produced outputs that were identical to the input or showed very minimal variation. This limited its effectiveness in generating meaningful paraphrases.

To improve performance, the model was switched to FLAN-T5 small, which is an instruction-tuned version of T5. This change resulted in better handling of prompts and slightly improved paraphrasing quality.

---

### Model Output Limitations

Even after switching models, the paraphrased outputs were sometimes:
- Too similar to the original text
- Overly simplified, leading to loss of context
- Inconsistent in structure for complex sentences

This highlighted the limitation of lightweight transformer models in generating highly diverse paraphrases.

---

### Dependency and Environment Issues

While integrating additional features such as evaluation metrics, several dependency conflicts occurred:
- Errors related to incompatible versions of libraries (e.g., transformers and sentence-transformers)
- Import errors that disrupted the execution of the notebook

These issues required multiple environment resets and careful selection of lightweight libraries to ensure stability.

---

### Kernel Crashes and Memory Constraints

During experimentation with different models and libraries, the Jupyter/Colab kernel occasionally crashed or stopped responding. This was mainly due to:
- High memory usage by transformer models
- Loading multiple large libraries simultaneously

As a result, lightweight models and minimal dependencies were preferred to maintain smooth execution.

---

### Choice of Development Environment

Although Jupyter Notebook was initially considered, running complex transformer models locally caused instability and performance issues. Therefore, Google Colab was used as the primary environment due to:
- Better resource availability (RAM and GPU support)
- Easier package management
- Improved execution stability

---

### Evaluation Challenges

Implementing evaluation metrics such as BLEU and ROUGE presented challenges:
- Some metrics produced very low scores due to structural differences in paraphrased text
- Advanced semantic similarity libraries caused compatibility issues

To address this, simpler and more stable evaluation methods were adopted.

---

### Summary

These challenges highlight the practical difficulties of working with NLP models, including model limitations, environment setup issues, and dependency management. Overcoming these challenges required careful model selection, simplification of the approach, and iterative testing.
